```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("loan.csv.zip",compression='zip')
df.head()
```

```python
df.shape
```

```python
cols_to_drop = [
    'id',
    'memberId',
    'url',
    'acceptD',
    'creditPullD',
    'expD',
    'ils_exp_d',
    'listD',
    'reviewStatusD',
    'title',
    'desc',
    'emp_title',
    'msa',
    'zip_code',
    'reviewStatus',
    'serviceFeeRate',
    'annual_inc_joint',
    'dti_joint',
    'verified_status_joint',
    'revol_bal_joint',
    'sec_app_fico_range_low',
    'sec_app_fico_range_high',
    'sec_app_earliest_cr_line',
    'sec_app_inq_last_6mths',
    'sec_app_mort_acc',
    'sec_app_open_acc',
    'sec_app_revol_util',
    'sec_app_open_act_il',
    'sec_app_num_rev_accts',
    'sec_app_chargeoff_within_12_mths',
    'sec_app_collections_12_mths_ex_med',
    'sec_app_mths_since_last_major_derog'
]

df = df.drop(columns=cols_to_drop, errors='ignore')

print("Dataset shape after dropping columns:", df.shape)
```

```python
collections_statuses = [
    'In Grace Period',
    'Late (16-30 days)',
    'Late (31-120 days)',
    'Default',
    'Charged Off'
]

df = df[df['loan_status'].isin(collections_statuses)]
df.shape
```

```python
loan_features = [
    'loan_amnt',
    'funded_amnt',
    'term',
    'int_rate',
    'installment',
    'purpose',
    'application_type',
    'grade',
    'sub_grade'
]
borrower_features = [
    'annual_inc',
    'emp_length',
    'home_ownership',
    'verification_status',
    'addr_state',
    'dti'
]
credit_features = [
    'earliest_cr_line',
    'pub_rec',
    'pub_rec_bankruptcies',
    'tax_liens'
]
delinq_features = [
    'acc_now_delinq',
    'delinq_2yrs',
    'delinq_amnt',
    'mths_since_last_delinq',
    'collections_12_mths_ex_med',
    'chargeoff_within_12_mths',
    'num_accts_ever_120_pd',
    'num_tl_30dpd',
    'num_tl_90g_dpd_24m'
]
stress_features = [
    'revol_bal',
    'revol_util',
    'all_util',
    'bc_util',
    'avg_cur_bal',
    'tot_cur_bal',
    'total_bal_ex_mort',
    'total_bc_limit',
    'il_util',
    'percent_bc_gt_75'
]
account_features = [
    'open_acc',
    'total_acc',
    'mort_acc',
    'pct_tl_nvr_dlq'
]
activity_features = [
    'inq_last_6mths',
    'inq_last_12m',
    'inq_fi',
    'acc_open_past_24mths',
    'open_acc_6m',
    'open_rv_12m',
    'open_rv_24m',
    'open_il_12m',
    'open_il_24m'
]
targets = [
    'loan_status',
    'recoveries',
    'collection_recovery_fee'
]
```

```python
final_cols = (
    loan_features +
    borrower_features +
    credit_features +
    delinq_features +
    stress_features +
    account_features +
    activity_features +
    targets
)

df_final = df[final_cols].copy()
df_final.shape
```

```python
missing_pct = (df_final.isnull().mean().sort_values(ascending=False))
print(missing_pct)
```

```python
drop_high_missing = [
    'il_util',
    'all_util',
    'open_il_12m',
    'open_il_24m',
    'open_rv_12m',
    'open_rv_24m',
    'open_acc_6m',
    'inq_fi',
    'inq_last_12m'
]
df_final = df_final.drop(columns=drop_high_missing)
df_final.shape
```

```python
df_final.to_csv("reduced_lending_club.csv",index=False)
```
